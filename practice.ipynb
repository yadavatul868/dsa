```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def sortedArrayToBST(self, nums):
        """
        Converts a sorted array into a height-balanced binary search tree.

        Args:
        nums: A list of integers sorted in ascending order.

        Returns:
        The root of the newly created height-balanced BST.
        """
        
        def helper(left, right):
            """
            Recursive helper function that builds the tree from a subarray segment.
            """
            # Base case: If the left pointer crosses the right pointer,
            # it means this segment of the array is empty, so we return None.
            if left > right:
                return None
            
            # Find the middle index of the current segment. This element will become the root
            # of the current subtree, ensuring the tree remains balanced.
            mid = (left + right) // 2
            
            # Create a new TreeNode with the middle element's value.
            root = TreeNode(nums[mid])
            
            # Recursively call the helper function to build the left subtree
            # using the left half of the current array segment (elements before mid).
            root.left = helper(left, mid - 1)
            
            # Recursively call the helper function to build the right subtree
            # using the right half of the current array segment (elements after mid).
            root.right = helper(mid + 1, right)
            
            return root

        # Start the recursive process on the entire array (from index 0 to len-1).
        return helper(0, len(nums) - 1)
```

#### _Invert Binary Trees_

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def add_child(self, data):
        
        # check if data is already present
        if data == self.val:
            return
        
        # Check if data is less than root node value
        elif data < self.val:
            
            # If the left node is also a Tree Node, call this function recursively
            if self.left:
                self.left.add_child(data)

            # # If the left node is None, add the data node here
            else:
                self.left = TreeNode(data)
        
        # Check if data is more than root node value
        elif data > self.val:
            
            # If the right node is also a Tree Node, call this function recursively
            if self.right:
                self.right.add_child(data)
            
            # If the right node is None, add the data node here
            else:
                self.right = TreeNode(data)
                
    def inorder_traversal(self):

        elements = []

        # visit left subtree
        if self.left:
            elements += self.left.inorder_traversal()

        # visit root node
        elements.append(self.val)

        # visit right subtree
        if self.right:
            elements += self.right.inorder_traversal()

        return elements
    
    def search(self, data):

        # if data is equal to val, return true
        if data == self.val:
            return True
        
        # if data is less than val
        elif data < self.val:

            # if left is a subtree, call the method recursively
            if self.left:
                return self.left.search(data)
            
            # if left is None, data not found
            else:
                return False

        # if data is more than val
        elif data > self.val:
            # if right is a subtree, call the method recursively
            if self.right:
                return self.right.search(data)
            # if right is None, data not found
            else:
                return False



def build_tree(elements):

    root = TreeNode(elements[0])

    for i in range(1, len(elements)):
        root.add_child(elements[i])

    return root




```

#### _Valid Palindrome_

```python
class Solution:

    def isPalindrome(self, s):
        
        l, r = 0, (len(s)-1)

        while l < r:

            while ((l < r) and not self.alphaNumeric(s[l])):
                l = l + 1

            while ((r > l) and not self.alphaNumeric(s[r])):
                r = r - 1
                
            if s[l].lower() != s[r].lower():
                return False

            l, r = (l + 1), (r - 1)

        return True


    def alphaNumeric(self, c):
        return ((ord('A') <= ord(c) <= ord('Z')) or (ord('a') <= ord(c) <= ord('z')) or (ord('0') <= ord(c) <= ord('9')))


SolutionObj = Solution()
SolutionObj.isPalindrome('madam')
```

#### _Valid Anagram_

```python
class Solution:

    def isAnagram(self, s, t):

        # compare len of the strings
        if len(s) != len(t):
            return False

        # create hashmap of of sting s & t
        dict_s = {}
        dict_t = {}
        for i in range(len(s)):

            # hasmap of s
            if s[i] not in dict_s:
                dict_s[s[i]] = 1
            else:
                dict_s[s[i]] = dict_s[s[i]] + 1

            # hashmap of t
            if t[i] not in dict_t:
                dict_t[t[i]] = 1
            else:
                dict_t[t[i]] = dict_t[t[i]] + 1

        # count occurence of each alphabet
        for key in dict_s.keys():
            if key not in dict_t.keys():
                return False
            else:
                if dict_s[key] != dict_t[key]:
                    return False
        
        return True

        
SolutionObj = Solution()
SolutionObj.isAnagram('anagram', 'nagaram')

# another way to solve this is to first sort both string and compare ??
```

#### _Linked List Cycle_

```python
# Flyod's Tortoise and Hare Problem

# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None


class Solution:

    def createLinkedList(self, ls, pos):

        output = []

        output.append(ListNode(ls[0]))

        for i in range(1, len(ls)):

            output.append(ListNode(ls[i]))
            output[(i-1)].next = output[i]

        if pos:
            output[(len(ls)-1)].next = output[pos]
        else:
            output[(len(ls)-1)].next = None

        return output


    def hasCycle(self, head):
        
        slow, fast = head, head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                return True

        return False
        

# Flyod Cycle Detection ??



SolutionObj = Solution()
ls = [3,2,0,-4]
pos = 2
head = SolutionObj.createLinkedList(ls, pos)
SolutionObj.hasCycle(head[0])

```

#### _Reverse Linked List_

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:

    def createLinkedList(self, ls, pos=None):
        output = []
        output.append(ListNode(ls[0]))
        
        for i in range(1, len(ls)):
            output.append(ListNode(ls[i]))
            output[(i-1)].next = output[i]
        
        if pos:
            output[(len(ls)-1)].next = output[pos]
        
        return output[0]

        
    def reverseList(self, head):
        prev = None
        curr = head

        while curr:
            temp = curr.next
            curr.next = prev
            prev = curr
            curr = temp

        return prev

    


SolutionObj = Solution()
ls = [1,2,3,4,5]
head = SolutionObj.createLinkedList(ls)
print(head.val, head.next.val)
head_rev = SolutionObj.reverseList(head=head)
print(head_rev.val, head_rev.next.val)


        
        
```

#### _Majority Element_

```python
class Solution:

    def majorityElement(self, nums):
        
        count = {}
        maxCount, maxCountNum = 0, 0

        for i in nums:

            if i not in count.keys():
                count[i] = 1
            else:
                count[i] = count[i] + 1
            
            if count[i] > maxCount:
                maxCount = count[i]
                maxCountNum = i
            
        return maxCountNum




SolutionObj = Solution()
SolutionObj.majorityElement([2,2,1,1,1,2,2])

```

#### _Missing Element_

```python
class Solution:
    def missingNumber(self, nums):

        res = len(nums)

        for i in range(len(nums)):

            res = res + (i - nums[i])
        
        return res
    

SolutionObj = Solution()
SolutionObj.missingNumber([3,0,1])
```

#### _Reverse String_

```python
class Solution:
    def reverseString(self, s):
        
        l, r = 0, len(s)-1

        while l < r:

            t = s[l]
            s[l] = s[r]
            s[r] = t
            
            l = l + 1
            r = r - 1

        return s
        

SolutionObj = Solution()
SolutionObj.reverseString(["h","e","l","l","o"])
```

#### _Diameter of Binary Tree_

```python
# Definition for a binary tree node.

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    
    def add_child(self, data):
        
        # check if data is already present
        if data == self.val:
            return
        
        # Check if data is less than root node value
        elif data < self.val:
            
            # If the left node is also a Tree Node, call this function recursively
            if self.left:
                self.left.add_child(data)

            # # If the left node is None, add the data node here
            else:
                self.left = TreeNode(data)
        
        # Check if data is more than root node value
        elif data > self.val:
            
            # If the right node is also a Tree Node, call this function recursively
            if self.right:
                self.right.add_child(data)
            
            # If the right node is None, add the data node here
            else:
                self.right = TreeNode(data)

                

class Solution:

    def heightOfBinaryTree(self, root):
        
        if not root:
            return -1
        else:
            left_height = self.heightOfBinaryTree(root.left)
            right_height = self.heightOfBinaryTree(root.right)
        
        return 1 + max(left_height, right_height)


    def diameterOfBinaryTree(self, root):
        self.res = 0

        if not root:
            return -1
        else:
            left_height = self.diameterOfBinaryTree(root.left)
            right_height = self.diameterOfBinaryTree(root.right)

            diameter = (left_height + right_height + 2)
            self.res = max(self.res, diameter)
            
            return 1 + max(left_height, right_height)
        
```

#### _Sorted array to binary tree_

```python
# Definition for a binary tree node.

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:

    def sortedArrayToBST(self, nums):
        
        l, r = 0, (len(nums)-1)

        if l > r:
            return None
        else:
            m = (l + r) // 2
            root = TreeNode(nums[m])

            # for left subtree
            l_left, r_left = l, m-1
            if l_left > r_left:
                root.left = None
            else:
                nums_left_sub = nums[l_left:(r_left+1)]
                root.left = self.sortedArrayToBST(nums_left_sub)

            # for right subtree
            l_right, r_right = m+1, r
            if l_right > r_right:
                root.right = None
            else:
                nums_right_sub = nums[l_right:(r_right+1)]
                root.right = self.sortedArrayToBST(nums_right_sub)




    
```

```python
def pairSum(arr, target):

    left_p = 0
    right_p = len(arr)-1

    while left_p < right_p:
        
        pair_sum = arr[left_p] + arr[right_p]
        
        # if pair_sum == target, return 1
        if pair_sum == target:
            return 1
        # if pair_sum < target, left_p += 1
        elif pair_sum < target:
            left_p += 1
        # if pair_sum > target, right_p -= 1
        elif pair_sum > target:
            right_p -= 1

    return 0

pairSum([1, 2, 3, 4, 6, 8], 10)
```

```python
def removeDuplicate(arr):
    
    left_p = 0
    right_p = 1

    while right_p < len(arr):

        if arr[left_p] == arr[right_p]:
            right_p += 1

        else:
            left_p += 1
            k = arr[left_p]
            arr[left_p] = arr[right_p]
            arr[right_p] = k
            right_p += 1

    return arr[:(left_p+1)]

removeDuplicate([1,1,2,4,4,5,5])
```

```python

nums = [4, 5, 6, 7, 0, 1, 2]

def modified_binary_search(nums, target):

    left = 0
    right = len(nums) - 1

    while left <= right:

        mid = (left + right) // 2
        mid_val = nums[mid]

        if target == mid_val:
            return mid

        # check if the left half is sorted
        if nums[left] <= nums[mid]:
            ## if it is sorted
            if nums[left] <= target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1

        # check if right half is sorted
        else:
            ## if it is sorted
            if nums[mid] < target <= nums[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1
```

```python
def subsets(nums):

    result = []
    subset = []

    def dfs(i):

        if i >= len(nums):
            result.append(subset.copy())
            return

        # decides to include nums[i] in subset
        subset.append(nums[i])
        dfs(i + 1)

        # decides not to include nums[i] in subset
        subset.pop()
        dfs(i + 1)

    dfs(0)
    return result
    


```

```python
def insert(arr, temp):

    if len(arr) == 0:
        arr.append(temp)
        return

    # write base case here
    elif ((len(arr) != 0) & (arr[-1] <= temp)):
        arr.append(temp)
        return
    

    # remove last element and save it in temp var
    val = arr[len(arr)-1]
    arr.pop()

    # recurrsively call sort function again
    insert(arr, temp)

    # write induction here
    arr.append(val)

    return
    



def sort_stack(arr):

    # write base case here
    if len(arr) <= 0:
        return

    # remove last element and save it in temp var
    temp = arr[len(arr)-1]
    arr.pop()

    # recurrsively call sort function again
    sort_stack(arr)

    # write induction here
    insert(arr, temp)

    return

```

```python
%%writefile kth_symbol_grammer.py


def kth_symbol(N, k):


    # base condition
    if ((N == 1) & (k == 1)):
        return 0

    # otherwise calculate length first
    c = pow(2, (N-1))
    mid = int(c / 2)

    # now recurrsive call
    if k <= mid:
        return kth_symbol(N-1, k)
    else:
        return 0 if kth_symbol(N-1, k - mid) == 1 else 1


    
```

```python
%%writefile tower_of_hanoi.py

def tower_hanoi(n, s, d, h):

    # base condition
    if n == 1:
        print('moving plate from {} to {}'.format(s, d))
        return

    # recurrsive call
    tower_hanoi((n-1), s, h, d)

    # induction step
    print('moving plate from {} to {}'.format(s, d))
    tower_hanoi((n-1), h, d, s)

    return

    
```
